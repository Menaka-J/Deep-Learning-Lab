**AIM:**
To implement a feed forward neural network for multiclass classification using the standard Iris dataset and train it using backpropagation.

```python
import numpy as np
from sklearn.datasets import load_iris
```

```python
iris = load_iris()

X = iris.data[:100]
y = iris.target[:100].reshape(-1, 1)

X = X / np.max(X, axis=0)
```

```python
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def sigmoid_derivative(x):
    return x * (1 - x)
```

```python
np.random.seed(1)

W1 = np.random.rand(4, 5)
b1 = np.zeros((1, 5))

W2 = np.random.rand(5, 1)
b2 = np.zeros((1, 1))
```

```python
epochs = 1000
lr = 0.1

for epoch in range(epochs):

    # Forward pass
    z1 = np.dot(X, W1) + b1
    a1 = sigmoid(z1)

    z2 = np.dot(a1, W2) + b2
    y_pred = sigmoid(z2)

    # Error
    error = y - y_pred
    loss = np.mean(error ** 2)

    # Backpropagation
    d_output = error * sigmoid_derivative(y_pred)
    d_hidden = d_output.dot(W2.T) * sigmoid_derivative(a1)

    # Update weights
    W2 += a1.T.dot(d_output) * lr
    b2 += np.sum(d_output, axis=0, keepdims=True) * lr

    W1 += X.T.dot(d_hidden) * lr
    b1 += np.sum(d_hidden, axis=0, keepdims=True) * lr

    if epoch % 100 == 0:
        print(f"Epoch {epoch}, Loss = {loss:.4f}")
```

```python
predicted = (y_pred >= 0.5).astype(int)
accuracy = np.mean(predicted == y) * 100

final_error = np.mean(np.abs(error))

print("\nFinal Results")
print("-------------")
print("Final Loss (MSE):", loss)
print("Final Error     :", final_error)
print("Accuracy        :", accuracy, "%")
print("Predicted values (first 10):")
print(predicted[:10].T)
```

**RESULT:**
The feed forward neural network successfully classified the Iris dataset into three classes with decreasing loss over epochs and achieved good classification accuracy.
