**AIM:**
   To implement a feed forward neural network for binary classification using synthetically generated data and train it using backpropagation.

```python
import numpy as np
```

```python
def generate_data(samples=200, noise_level=0.5):
    np.random.seed(42)
    X = np.random.randn(samples, 2)
    X += noise_level * np.random.randn(samples, 2)

    y = np.zeros((samples, 1))
    for i in range(samples):
        if X[i].sum() > 0:
            y[i] = 1
        else:
            y[i] = 0

    return X, y
```

```python
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def sigmoid_derivative(x):
    return x * (1 - x)

def mse_loss(y, y_pred):
    return np.mean((y - y_pred) ** 2)
```

```python
X, y = generate_data(samples=200)
```

```python
np.random.seed(1)

W1 = np.random.randn(2, 5)
b1 = np.zeros((1, 5))

W2 = np.random.randn(5, 1)
b2 = np.zeros((1, 1))
```

```python
epochs = 1500
lr = 0.05

for epoch in range(epochs):

    # ---- Forward Pass ----
    z1 = np.dot(X, W1) + b1
    a1 = sigmoid(z1)

    z2 = np.dot(a1, W2) + b2
    y_pred = sigmoid(z2)

    # ---- Loss ----
    loss = mse_loss(y, y_pred)

    # ---- Backpropagation ----
    error = y - y_pred
    d_output = error * sigmoid_derivative(y_pred)
    d_hidden = d_output.dot(W2.T) * sigmoid_derivative(a1)

    # ---- Update ----
    W2 += a1.T.dot(d_output) * lr
    b2 += np.sum(d_output, axis=0, keepdims=True) * lr

    W1 += X.T.dot(d_hidden) * lr
    b1 += np.sum(d_hidden, axis=0, keepdims=True) * lr

    # ---- Print Loss ----
    if epoch % 100 == 0:
        print(f"Epoch {epoch}, Loss = {loss:.4f}")
```

```python
predicted_classes = (y_pred >= 0.5).astype(int)
accuracy = np.mean(predicted_classes == y) * 100

final_error = np.mean(np.abs(error))

print("\nFinal Results")
print("-------------")
print("Final Error     :", final_error)
print("Final Loss:", loss)
print("Accuracy:", accuracy, "%")
print("Predicted classes (first 10):")
print(predicted_classes[:10].T)
```

**RESULT:**
    The feed forward neural network successfully classified the synthetic data with decreasing loss over epochs and achieved good classification accuracy.
